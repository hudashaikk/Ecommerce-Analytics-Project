```python
# Convert date columns from text to datetime
df["Order Date"] = pd.to_datetime(df["Order Date"], dayfirst=True)
df["Ship Date"] = pd.to_datetime(df["Ship Date"], dayfirst=True)

# Check the result
print(df[["Order Date", "Ship Date"]].dtypes)
```

```python
# Create date features

df["Year"] = df["Order Date"].dt.year
df["Month"] = df["Order Date"].dt.month
df["Month Name"] = df["Order Date"].dt.month_name()
df["Quarter"] = df["Order Date"].dt.quarter

# Calculate shipping time
df["Shipping Days"] = (
    df["Ship Date"] - df["Order Date"]
).dt.days

# Display the result
df[[
    "Order Date",
    "Ship Date",
    "Year",
    "Month",
    "Month Name",
    "Quarter",
    "Shipping Days"
]].head()
```

```python
# Create a clean copy of the dataset
clean_df = df.copy()

# Check the cleaned dataset
print("Rows:", clean_df.shape[0])
print("Columns:", clean_df.shape[1])
print("Missing values:")
print(clean_df.isnull().sum())
```

5. Yearly Sales Analysis


```python
# Calculate total sales by year

yearly_sales = (
    clean_df.groupby("Year")["Sales"]
    .sum()
    .reset_index()
)

print(yearly_sales)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 5))

plt.plot(
    yearly_sales["Year"],
    yearly_sales["Sales"],
    marker="o"
)

plt.title("Total Sales by Year")
plt.xlabel("Year")
plt.ylabel("Total Sales")
plt.grid(True)

plt.show()
```

```python
yearly_sales["YoY Growth %"] = (
    yearly_sales["Sales"].pct_change() * 100
)

yearly_sales
```

## Yearly Sales Analysis

The analysis shows that sales declined by 4.26% in 2016 compared with 2015.
Sales then increased by 30.64% in 2017 and by 20.30% in 2018.

The highest annual sales were recorded in 2018, at approximately $722,052.

6. Monthly Sales Analysis


```python
monthly_sales = (
    clean_df.groupby(["Year", "Month"])["Sales"]
    .sum()
    .reset_index()
)

monthly_sales.head(12)
```

```python
monthly_sales["Year-Month"] = pd.to_datetime(
    monthly_sales["Year"].astype(str) + "-" +
    monthly_sales["Month"].astype(str) + "-01"
)

monthly_sales
```

```python
plt.figure(figsize=(12, 5))

plt.plot(
    monthly_sales["Year-Month"],
    monthly_sales["Sales"],
    marker="o"
)

plt.title("Monthly Sales Trend")
plt.xlabel("Date")
plt.ylabel("Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.show()
```

```python
top_months = monthly_sales.sort_values(
    "Sales",
    ascending=False
).head(10)

top_months[["Year", "Month", "Sales"]]
```

```python
lowest_months = monthly_sales.sort_values(
    "Sales",
    ascending=True
).head(10)

lowest_months[["Year", "Month", "Sales"]]
```

```python
monthly_seasonality = (
    clean_df.groupby("Month")["Sales"]
    .agg(["sum", "mean", "count"])
    .reset_index()
)

monthly_seasonality
```

```python
monthly_seasonality.sort_values(
    "mean",
    ascending=False
)
```

## Monthly Sales Seasonality

Monthly analysis shows variation in average sales across the year.

March has the highest average sales per record at approximately $290.55,
while February has the lowest average at approximately $199.90.

January, October, November, and December also show relatively high
average sales compared with several other months.

These patterns indicate variation in sales performance across months,
which can be investigated further by category, product, and region.

7. Category Analysis


```python
#Which product categories generate the most revenue?
category_sales = (
    clean_df.groupby("Category")["Sales"]
    .agg(["sum", "mean", "count"])
    .reset_index()
)

category_sales = category_sales.sort_values(
    "sum",
    ascending=False
)

category_sales
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    category_sales["Category"],
    category_sales["sum"]
)

plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Total Sales")

plt.show()
```

```python
category_sales
```

```python
category_sales["Sales Share %"] = (
    category_sales["sum"] /
    category_sales["sum"].sum()
) * 100

category_sales[
    ["Category", "sum", "Sales Share %"]
]
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    category_sales["Category"],
    category_sales["Sales Share %"]
)

plt.title("Sales Contribution by Category")
plt.xlabel("Category")
plt.ylabel("Sales Share (%)")

plt.show()
```

## Category Sales Analysis

Technology generated the largest share of total sales at approximately 36.6%.
Furniture contributed approximately 32.2%, while Office Supplies contributed
approximately 31.2%.

The sales contribution is relatively balanced across the three categories,
with no single category accounting for a majority of total sales.

9. Product Analysis


```python
#Which products generate the most revenue?

product_sales = (
    clean_df.groupby("Product Name")["Sales"]
    .agg(["sum", "mean", "count"])
    .reset_index()
)

product_sales = product_sales.sort_values(
    "sum",
    ascending=False
)

top_10_products = product_sales.head(10)

top_10_products
```

```python
#Because product names can be long, use a horizontal bar chart:
top_10_products = top_10_products.sort_values(
    "sum",
    ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
    top_10_products["Product Name"],
    top_10_products["sum"]
)

plt.title("Top 10 Products by Total Sales")
plt.xlabel("Total Sales")
plt.ylabel("Product")

plt.show()
```

```python
bottom_10_products = product_sales.sort_values(
    "sum",
    ascending=True
).head(10)

bottom_10_products
```

```python
#Let's calculate how much of the company's total sales come from the top 10 products:
total_sales = clean_df["Sales"].sum()

top_10_sales = top_10_products["sum"].sum()

top_10_share = (top_10_sales / total_sales) * 100

print("Total Sales:", total_sales)
print("Top 10 Product Sales:", top_10_sales)
print("Top 10 Sales Share:", top_10_share, "%")
```

10. Customer Analysis

```python
# Customer Analysis

customer_sales = (
    df.groupby("Customer Name")
      .agg(
          Total_Sales=("Sales", "sum"),
          Order_Count=("Order ID", "nunique")
      )
      .sort_values("Total_Sales", ascending=False)
)

customer_sales.head(10)
```

```python
# Top 10 customers by sales

top_customers = customer_sales.head(10)

plt.figure(figsize=(10, 5))

plt.bar(
    top_customers.index,
    top_customers["Total_Sales"]
)

plt.title("Top 10 Customers by Sales")
plt.xlabel("Customer")
plt.ylabel("Total Sales")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y")

plt.show()
```

```python
segment_sales = (
    df.groupby("Segment")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

segment_sales
```

```python
plt.figure(figsize=(8, 5))

plt.bar(
    segment_sales.index,
    segment_sales.values
)

plt.title("Total Sales by Customer Segment")
plt.xlabel("Segment")
plt.ylabel("Total Sales")
plt.grid(axis="y")

plt.show()
```

12. Shipping Analysis

We want to answer:

Which shipping mode is used most?
Which shipping mode generates the most sales?
How many days do orders typically take to ship?
Is there a noticeable difference between shipping modes?

```python
#1. Orders by shipping mode
shipping_orders = (
    df.groupby("Ship Mode")["Order ID"]
      .nunique()
      .sort_values(ascending=False)
)

shipping_orders
```

```python
#sales by shipping mode
shipping_sales = (
    df.groupby("Ship Mode")["Sales"]
      .sum()
      .sort_values(ascending=False)
)

shipping_sales
```

```python
#3. Average shipping time
shipping_time = (
    df.groupby("Ship Mode")["Shipping Days"]
      .agg(["mean", "median", "count"])
      .sort_values("mean")
)

shipping_time
```

```python
#4. Visualize average shipping time
plt.figure(figsize=(8, 5))

plt.bar(
    shipping_time.index,
    shipping_time["mean"]
)

plt.title("Average Shipping Time by Ship Mode")
plt.xlabel("Ship Mode")
plt.ylabel("Average Shipping Days")
plt.grid(axis="y")

plt.show()
```

```python
#Let's calculate the overall average shipping time:
overall_shipping_days = df["Shipping Days"].mean()

print(
    f"Average shipping time: {overall_shipping_days:.2f} days"
)
```

13. Final Business KPIs

```python
# Final Business KPIs

total_sales = clean_df["Sales"].sum()
total_orders = clean_df["Order ID"].nunique()
total_customers = clean_df["Customer ID"].nunique()
total_products = clean_df["Product ID"].nunique()
average_order_value = total_sales / total_orders
average_shipping_days = clean_df["Shipping Days"].mean()

print(f"Total Sales: ${total_sales:,.2f}")
print(f"Total Orders: {total_orders:,}")
print(f"Total Customers: {total_customers:,}")
print(f"Total Products: {total_products:,}")
print(f"Average Order Value: ${average_order_value:,.2f}")
print(f"Average Shipping Time: {average_shipping_days:.2f} days")
```
